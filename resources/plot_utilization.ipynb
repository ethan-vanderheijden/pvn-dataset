```python
import pandas as pd
import seaborn as sns

DATA_FILE = "combined.csv"
FIELD = "cpu"
FILTER_OUT_SYSTEM = "systemwide"
```

```python
# Timeseries
data = pd.read_csv(DATA_FILE)

if FILTER_OUT_SYSTEM:
    data = data[data["system"] != FILTER_OUT_SYSTEM]
# data["cpu"] = data["cpu"] / 100

plot = sns.lineplot(data=data, x="time", y=FIELD, hue="system")
plot.set_xlabel("Time (s)")
```

```python
# Histogram
sns.histplot(data=data, x=FIELD, hue="system", element="bars", stat="probability", common_norm=False, binrange=(0, 100), binwidth=5)
```
