```python
import pandas as pd
import seaborn as sns

DATA_FILE = "temp.csv"
FIELD = "cpu"
FILTER_SYSTEM = "testing"
```

```python
# Timeseries
data = pd.read_csv(DATA_FILE)

if FILTER_SYSTEM:
    data = data[data["system"] == FILTER_SYSTEM]
# data["cpu"] = data["cpu"] / 100

plot = sns.lineplot(data=data, x="time", y=FIELD, hue="system")
plot.set_xlabel("Time (s)")
```

```python
# Histogram
sns.histplot(data=data, x=FIELD, hue="system", element="bars", stat="probability", common_norm=False)
```
