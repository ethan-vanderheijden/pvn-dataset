```python
import pandas as pd
import seaborn as sns

DATA_FILE = "utilization_data/rdr_multiplexed.csv"
FIELD = "cpu"
```

```python
# Timeseries
data = pd.read_csv(DATA_FILE)

data["cpu"] = data["cpu"] / 100

plot = sns.lineplot(data=data, x="time", y=FIELD, hue="system")
plot.set_xlabel("Time (s)")
```

```python
# Histogram
sns.histplot(data=data, x=FIELD, hue="system", element="step", stat="probability", common_norm=False)
```
