```python
import pandas as pd
import seaborn as sns

DATA_FILE = "utilization_data/transcoder_2.csv"
FIELD = "cpu"
```

```python
# Timeseries
data = pd.read_csv(DATA_FILE)
sns.lineplot(data=data, x="time", y=FIELD, hue="system")
```

```python
# Histogram
sns.histplot(data=data, x=FIELD, hue="system", element="step", stat="probability", common_norm=False)
```
