```python
import pandas as pd
import seaborn as sns
import math
```

```python
data = pd.read_csv("perf-scheduling.csv")
data["time"] = data["time"] / 60
```

```python
ax = sns.lineplot(data=data, x="time", y="transcode_duration", hue="scheduler")
ax.set_xlabel("Time (m)")
max_minute = math.ceil(data["time"].max())
ax.set_xticks(range(0, max_minute + 1, 3))
ax.set_ylabel("Transcode Duration (s)")
ax.margins(x=0)

labels = ["0%", "20%", "40%", "60%", "80%", "100%", "200%"]
width_in_secs = 3
for i, label in enumerate(labels):
    ax.axvspan(
        i * width_in_secs, (i + 1) * width_in_secs, label=label, alpha=0.1 * i, color="gray"
    )
    ax.text(
        (i + 0.5) * width_in_secs,
        15.5,
        label,
        horizontalalignment="center",
        verticalalignment="center",
    )
```
